```python
import numpy as np
import matplotlib.pyplot as plt

E = np.linspace(-0.5, 0.5, 1000)
Ef = 0.0
kB = 8.617333e-5
T = [100, 300, 600, 1000]

for temp in T:

    fE = 1 / (1 + np.exp((E - Ef) / (kB * temp)))
    plt.plot(E, fE, label=f"T = {temp} K")

plt.axvline(Ef, linestyle="--", label="Fermi Level")
plt.legend()
plt.title("Fermi function")
plt.xlabel("E")
plt.ylabel("f(E)")

plt.grid()
plt.show()

```
